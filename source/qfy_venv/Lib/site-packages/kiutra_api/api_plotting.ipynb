```python
import matplotlib.pyplot as plt
import numpy as np
import time
from kiutra_api.device_interfaces import Thermometer
from kiutra_api.controller_interfaces import MagnetControl
from datetime import datetime
from collections import deque
%matplotlib notebook


interval = 1
fig, ax = plt.subplots(2, 1, figsize=(7, 9))
plt.ion()

n_entries = 3600
ts = deque(maxlen=n_entries)
Ts = deque(maxlen=n_entries)
mag_ts = deque(maxlen=n_entries)
Bs = deque(maxlen=n_entries)

host = '192.168.11.20'
T_sample = Thermometer('T_sample', host=host)
sample_magnet = MagnetControl('sample_magnet', host)


while True:
    T, t = T_sample.kelvin_timestamp
    B, mag_t = sample_magnet.tesla_timestamp
    ts.append(datetime.fromtimestamp(t))
    Ts.append(T)
    mag_ts.append(datetime.fromtimestamp(mag_t))
    Bs.append(B)

    stable = sample_magnet.stable
    _, status = sample_magnet.status

    print(f'T = {T:.3f} K, B = {B:.3f} {"T":<2} ({status}, stable={stable})', end='\r')

    ax[0].clear()
    ax[0].plot(ts, Ts, label='T_sample')
    ax[0].set(ylabel='Temperature (K)')
    ax[1].clear()
    ax[1].plot(ts, Bs, label='B_sample')
    ax[1].set(ylabel='Field (T)')

    fig.canvas.draw()
    time.sleep(1)
```
